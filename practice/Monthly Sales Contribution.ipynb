```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

electronics = [20, 25, 22, 30, 35, 40]
clothing = [15, 18, 20, 22, 25, 28]
groceries = [10, 12, 15, 14, 18, 20]
```

```python
plt.figure(figsize=(9, 5))

plt.stackplot(
    months,
    electronics,
    clothing,
    groceries,
    labels=["Electronics", "Clothing", "Groceries"],
    alpha=0.7
)

plt.title("Monthly Sales Contribution")
plt.xlabel("Months")
plt.ylabel("Sales (in thousands)")
plt.legend(loc="upper left")

plt.show()
```

```python
print("Total Electronics Sales:", sum(electronics))
print("Total Clothing Sales:", sum(clothing))
print("Total Groceries Sales:", sum(groceries))
```
