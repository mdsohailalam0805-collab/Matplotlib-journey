```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

patna = [32, 34, 33, 35, 36, 34, 33]
delhi = [30, 32, 35, 36, 34, 33, 31]
```

```python
plt.xlabel("Days", color = "green")
plt.ylabel("Temperature (°C)", color = "green")
plt.title("Weekly Temperature Comparison", color = "red")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
print("Patna Average:", round(sum(patna) / len(patna), 2))
print("Delhi Average:", round(sum(delhi) / len(delhi), 2))

print("Highest Patna Temperature:", max(patna))
print("Highest Delhi Temperature:", max(delhi))
```
