```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

laptops = [40, 50, 45, 60, 70]
mobiles = [55, 60, 65, 70, 80]
tablets = [20, 25, 30, 35, 40]
```

```python
x = np.arange(len(months))

plt.figure(figsize=(9, 5))

plt.bar(x, laptops, label="Laptops")
plt.bar(x, mobiles, bottom=laptops, label="Mobiles")

bottom_values = np.array(laptops) + np.array(mobiles)

plt.bar(x, tablets, bottom=bottom_values, label="Tablets")

plt.xticks(x, months)

plt.title("Monthly Product Sales")
plt.xlabel("Months")
plt.ylabel("Units Sold")


plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

print("Total Sales:", sum(laptops) + sum(mobiles) + sum(tablets))
```
