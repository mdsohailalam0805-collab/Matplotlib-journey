```python
import matplotlib.pyplot as plt
```

```python
products = ["Laptop", "Mobile", "Tablet", "Smartwatch", "Headphones"]
prices = [55000, 25000, 18000, 5000, 3000]
```

```python
plt.figure(figsize=(9, 5))

plt.barh(products, prices, color="skyblue", edgecolor="black")

plt.xlabel("Price (₹)")
plt.ylabel("Products")
plt.title("Product Price Comparison")
plt.grid(axis="x", linestyle="--", alpha=0.4)

plt.show()
```

```python
print("Average Price:", round(sum(prices) / len(prices), 2))
print("Most Expensive:", products[prices.index(max(prices))])
print("Cheapest Product:", products[prices.index(min(prices))])
print("Total Price:", sum(prices))
```

Short explanation
- plt.barh() — horizontal bar chart banata hai.
- products — Y-axis par product names dikhata hai.
- prices — X-axis par product prices dikhata hai.
- max() — sabse expensive product find karta hai.
- min() — sabse cheap product find karta hai.
