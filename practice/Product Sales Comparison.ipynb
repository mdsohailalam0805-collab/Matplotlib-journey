```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]

laptop_sales = [40, 50, 45, 60, 70]
mobile_sales = [55, 60, 65, 70, 80]

x = np.arange(len(months))
width = 0.35

plt.figure(figsize=(9, 5))

plt.bar(x - width/2, laptop_sales, width, label="Laptops")
plt.bar(x + width/2, mobile_sales, width, label="Mobiles")

plt.xticks(x, months)

plt.title("Monthly Product Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Units Sold")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

print("Total Laptop Sales:", sum(laptop_sales))
print("Total Mobile Sales:", sum(mobile_sales))
```
