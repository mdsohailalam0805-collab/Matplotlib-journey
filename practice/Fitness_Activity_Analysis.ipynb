```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
steps = [4500, 7000, 6200, 8500, 5000, 11000, 9500]
```

```python
plt.figure(figsize=(9, 5))

plt.plot(days, steps, marker="o", linewidth=2, label="Daily Steps")
plt.axhline(8000, linestyle="--", label="Step Goal")

plt.xlabel("Days")
plt.ylabel("Number of Steps")
plt.title("Weekly Fitness Activity")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
print("Total Steps:", sum(steps))
print("Average Daily Steps:", round(sum(steps) / len(steps)))
print("Maximum Steps:", max(steps))
print("Most Active Day:", days[steps.index(max(steps))])
print("Days Goal Achieved:", sum(step >= 8000 for step in steps))
```
