```python
import matplotlib.pyplot as plt
from statistics import median
```

```python
it = [35000, 40000, 42000, 45000, 48000, 52000, 90000]
hr = [28000, 30000, 32000, 35000, 37000, 40000, 42000]
sales = [25000, 27000, 30000, 33000, 38000, 45000, 70000]

departments = [it, hr, sales]
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    departments,
    labels=["IT", "HR", "Sales"]
)

plt.title("Department-wise Salary Distribution")
plt.xlabel("Departments")
plt.ylabel("Salary (₹)")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
print("IT Median Salary:", median(it))
print("HR Median Salary:", median(hr))
print("Sales Median Salary:", median(sales))
```
