```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
sales = [5000, 7000, 4500, 9000, 8000, 12000, 10000]
orders = [10, 14, 9, 18, 16, 25, 20]
```

```python
plt.figure(figsize=(10, 5))

plt.plot(days, sales, marker="o", label="Sales (₹)")
plt.plot(days, [x * 500 for x in orders], marker="s", label="Orders × 500")

plt.xlabel("Days")
plt.ylabel("Amount / Scaled Orders")
plt.title("Weekly E-commerce Performance")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
print("Total Sales: ₹", sum(sales))
print("Total Orders:", sum(orders))
print("Average Daily Sales: ₹", round(sum(sales) / len(sales), 2))
print("Best Sales Day:", days[sales.index(max(sales))])
print("Highest Orders Day:", days[orders.index(max(orders))])
```
