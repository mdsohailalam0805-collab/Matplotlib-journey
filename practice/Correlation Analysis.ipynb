```python
import matplotlib.pyplot as plt
```

```python
advertising = [1000, 2000, 3000, 4000, 5000, 6000, 7000, 8000]
sales = [10, 18, 25, 30, 42, 48, 55, 65]
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(advertising, sales, s=100, marker="o")

plt.xlabel("Advertising Budget (₹)")
plt.ylabel("Sales (Units)")
plt.title("Advertising vs Sales Analysis")
plt.grid(True, alpha=0.4)

plt.show()
```

```python
print("Average Advertising:", sum(advertising) / len(advertising))
print("Average Sales:", round(sum(sales) / len(sales), 2))
print("Highest Sales:", max(sales))
```
