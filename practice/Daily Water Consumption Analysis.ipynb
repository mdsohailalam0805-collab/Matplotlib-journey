```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
water = [1.5, 2.0, 1.8, 2.5, 2.2, 3.0, 2.4]
```

```python
plt.plot(days, water, marker="o", linewidth=2, label="Water Intake")
plt.axhline(2, linestyle="--", label="2-Litre Target")
plt.xlabel("Days")
plt.ylabel("Water (Litres)")
plt.title("Weekly Water Consumption")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
