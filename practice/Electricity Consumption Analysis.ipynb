```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
units = [12, 15, 10, 18, 14, 22, 20]
```

```python
plt.bar(days, units, color="orange", edgecolor="black")

plt.axhline(sum(units) / len(units), linestyle="--", label="Average Usage")

plt.xlabel("Days")
plt.ylabel("Electricity (kWh)")
plt.title("Weekly Electricity Consumption")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
print("Total Consumption:", sum(units), "kWh")
print("Average Consumption:", round(sum(units) / len(units), 2), "kWh")
print("Maximum Consumption:", max(units), "kWh")
print("Highest Usage Day:", days[units.index(max(units))])
```
