```python
import matplotlib.pyplot as plt
```

```python
students = ["Aman", "Rahul", "Sohail", "Priya", "Neha", "Riya"]
attendance = [85, 70, 92, 65, 78, 88]
```

```python
plt.bar(students, attendance, edgecolor="black")

plt.axhline(75, linestyle="--", label="Minimum Attendance (75%)")

plt.xlabel("Students")
plt.ylabel("Attendance (%)")
plt.title("Student Attendance Analysis")
plt.ylim(0, 100)
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```
