```python
import matplotlib.pyplot as plt
```

```python
students = ["Faiz", "Saif", "Sohail", "Altamash", "Alqamar", "Shahjain"]
attendance = [95, 100, 50, 80, 78, 90]
```

```python
plt.bar(students, attendance, edgecolor="black")

plt.axhline(75, linestyle="--", label="Minimum Attendance (75%)")

plt.xlabel("Students")
plt.ylabel("Attendance (%)")
plt.title("Student Attendance Analysis")
plt.ylim(0, 100)
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
eligible = [students[i] for i in range(len(students)) if attendance[i] >= 75]

print("Average Attendance:", round(sum(attendance) / len(attendance), 2), "%")
print("Eligible Students:", eligible)
print("Students Below 75%:", sum(value < 75 for value in attendance))
```
