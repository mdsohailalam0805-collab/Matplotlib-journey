```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = np.array([10000, 12000, 15000, 13000, 18000, 22000])

# Calculate month-to-month percentage change
growth = np.diff(sales) / sales[:-1] * 100

growth_months = months[1:]

plt.figure(figsize=(9, 5))

plt.bar(
    growth_months,
    growth,
    label="Sales Growth"
)

plt.axhline(0, color="black", linewidth=1)

plt.title("Monthly Sales Growth (%)")
plt.xlabel("Months")
plt.ylabel("Growth (%)")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

print("Monthly Growth (%):", np.round(growth, 2))
```
