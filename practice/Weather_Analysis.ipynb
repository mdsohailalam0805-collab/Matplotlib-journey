```python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [30, 32, 29, 34, 33, 31, 35]
```

```python
plt.plot(days, temperature, marker="o", linestyle="-", label="Temperature")

plt.xlabel("Days")
plt.ylabel("Temperature (°C)")
plt.title("Weekly Weather Analysis")
plt.grid(True)
plt.legend()
plt.show()
```

```python
print("Average Temperature:", round(sum(temperature) / len(temperature), 2), "°C")
```

```python
print("Highest Temperature:", max(temperature), "°C")
```

```python
print("Lowest Temperature:", min(temperature), "°C")
```

```python
print("Hottest Day:", days[temperature.index(max(temperature))])
```
