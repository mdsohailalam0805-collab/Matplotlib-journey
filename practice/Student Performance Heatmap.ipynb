```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
students = ["Rehan", "Sohail", "Saif", "Alatmash"]
subjects = ["Python", "SQL", "ML", "Statistics"]

marks = np.array([
    [85, 78, 90, 80],
    [70, 88, 75, 82],
    [95, 92, 89, 94],
    [60, 65, 70, 68]
])
```

```python
plt.figure(figsize=(8, 5))

plt.imshow(marks, cmap="YlGnBu", aspect="auto")

plt.colorbar(label="Marks")

plt.xticks(range(len(subjects)), subjects)
plt.yticks(range(len(students)), students)

# Display marks inside each cell
for i in range(len(students)):
    for j in range(len(subjects)):
        plt.text(j, i, marks[i, j], ha="center", va="center")
        
plt.title("Student Performance Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")

plt.show()
```

```python
print("Class Average:", round(np.mean(marks), 2))
```
