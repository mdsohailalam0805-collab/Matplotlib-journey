```python

import matplotlib.pyplot as plt
```

```python

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

profit = [5000, 7000, 6000, 9000, 12000, 15000]
target = [6000, 6000, 6000, 8000, 10000, 10000]
```

```python
plt.figure(figsize=(9, 5))

plt.plot(months, profit, marker="o", label="Actual Profit")
plt.plot(months, target, linestyle="--", label="Target Profit")

plt.fill_between(
    months,
    profit,
    target,
    alpha=0.2
)
```
