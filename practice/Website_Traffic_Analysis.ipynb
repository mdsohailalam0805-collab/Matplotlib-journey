```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
visitors = [120, 150, 100, 180, 220, 300, 250]
page_views = [300, 400, 250, 500, 600, 850, 700]
```

```python
plt.figure(figsize=(9, 5))

plt.plot(days, visitors, marker="o", label="Visitors")
plt.plot(days, page_views, marker="s", label="Page Views")

plt.xlabel("Days")
plt.ylabel("Count")
plt.title("Weekly Website Traffic")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
print("Total Visitors:", sum(visitors))
print("Total Page Views:", sum(page_views))
print("Average Daily Visitors:", round(sum(visitors) / len(visitors), 2))
print("Highest Traffic Day:", days[visitors.index(max(visitors))])
```
