```python

import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [100, 100, 150, 150, 200, 250]
```

```python
plt.figure(figsize=(9, 5))

plt.step(
    months,
    sales,
    where="mid",
    marker="o",
    label="Monthly Sales"
)

plt.title("Monthly Sales Step Chart")
plt.xlabel("Months")
plt.ylabel("Sales (Units)")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
