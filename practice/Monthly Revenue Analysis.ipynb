```python

import matplotlib.pyplot as plt

departments = ["IT", "HR", "Sales", "Marketing"]

product_revenue = [40, 25, 60, 35]
service_revenue = [30, 35, 40, 45]

plt.figure(figsize=(9, 5))

plt.barh(
    departments,
    product_revenue,
    label="Product Revenue"
)

plt.barh(
    departments,
    service_revenue,
    left=product_revenue,
    label="Service Revenue"
)

plt.title("Department-wise Revenue Distribution")
plt.xlabel("Revenue (₹ Lakhs)")
plt.ylabel("Departments")

plt.legend()
plt.grid(axis="x", alpha=0.3)

plt.show()

total_revenue = [
    product_revenue[i] + service_revenue[i]
    for i in range(len(departments))
]

print("Total Revenue:", sum(total_revenue), "Lakhs")
print("Highest Revenue Department:",
    departments[total_revenue.index(max(total_revenue))])
```
