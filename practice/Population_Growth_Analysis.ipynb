```python
import matplotlib.pyplot as plt
```

```python
years = [2018, 2019, 2020, 2021, 2022, 2023, 2024]
population = [1000, 1100, 1250, 1300, 1450, 1600, 1750]
```

```python

plt.figure(figsize=(9, 5))

plt.plot(years, population, marker="o", linewidth=2)

plt.xlabel("Years")
plt.ylabel("Population")
plt.title("Population Growth Analysis")
plt.grid(True, linestyle="--", alpha=0.4)

plt.show()
```

```python
growth = population[-1] - population[0]

print("Starting Population:", population[0])
print("Latest Population:", population[-1])
print("Total Growth:", growth)
print("Average Population:", round(sum(population) / len(population), 2))
```
