```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
expenses = [5000, 6500, 4500, 7000, 6000, 8000]
```

```python
plt.figure(figsize=(12, 4))

# 1. Bar Chart
plt.subplot(1, 3, 1)
plt.bar(months, expenses)
plt.title("Monthly Expenses")
plt.xticks(rotation=45)
```
