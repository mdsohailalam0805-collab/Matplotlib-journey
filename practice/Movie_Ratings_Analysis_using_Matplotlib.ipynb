```python
import matplotlib.pyplot as plt

movies = ["Avatar", "Titanic", "Inception", "Interstellar", "Joker"]
ratings = [7.9, 7.9, 8.8, 8.7, 8.4]
```

```python
plt.bar(movies, ratings)
plt.axhline(sum(ratings) / len(ratings), linestyle="--", label="Average Rating")



plt.xlabel("Movies")
plt.ylabel("Rating")
plt.title("Movie Ratings Analysis")
plt.legend()
plt.xticks(rotation=20)
plt.show()

print("Average Rating:", round(sum(ratings) / len(ratings), 2))
print("Highest Rated Movie:", movies[ratings.index(max(ratings))])
```
