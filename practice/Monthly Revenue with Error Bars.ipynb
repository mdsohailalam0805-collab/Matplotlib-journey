```python

import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

revenue = [12000, 15000, 14000, 18000, 22000, 25000]
variation = [1000, 1500, 1200, 1800, 2000, 1500]
```

```python
plt.figure(figsize=(9, 5))

plt.errorbar(
    months,
    revenue,
    yerr=variation,
    fmt="o",
    capsize=5,
    color="blue"
)
```
