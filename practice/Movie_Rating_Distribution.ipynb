```python
import matplotlib.pyplot as plt
```

```python
ratings = [7.5, 8.2, 6.8, 9.0, 7.9, 8.5, 6.5, 7.2,
8.8, 9.2, 7.0, 6.9, 8.1, 7.7, 8.4, 9.1]
```

```python
plt.hist(ratings, bins=5, edgecolor="black")

plt.xlabel("Movie Ratings")
plt.ylabel("Number of Movies")
plt.title("Movie Rating Distribution")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
print("Total Movies:", len(ratings))
print("Average Rating:", round(sum(ratings) / len(ratings), 2))
print("Highest Rating:", max(ratings))
print("Lowest Rating:", min(ratings))
```
