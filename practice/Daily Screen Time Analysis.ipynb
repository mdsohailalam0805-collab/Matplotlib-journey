```python
import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
screen_time = [3, 4, 2.5, 5, 4.5, 7, 6]
```

```python
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.plot(days, screen_time, marker="o")
plt.title("Daily Screen Time")
plt.xlabel("Days")
plt.ylabel("Hours")
plt.grid(True, linestyle="--", alpha=0.5)



plt.subplot(1, 2, 2)
plt.scatter(days, screen_time, s=100)
plt.title("Screen Time Distribution")
plt.xlabel("Days")
plt.ylabel("Hours")

```
