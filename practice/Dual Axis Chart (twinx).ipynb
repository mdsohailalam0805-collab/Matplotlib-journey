```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

visitors = [1200, 1500, 1800, 2200, 2500, 3000]
conversion_rate = [2.5, 3.0, 2.8, 3.5, 4.0, 4.5]
```

```python
fig, ax1 = plt.subplots(figsize=(9, 5))

# First Y-axis: Visitors
ax1.plot(months, visitors, marker="o", label="Visitors")
ax1.set_xlabel("Months")
ax1.set_ylabel("Visitors")



# Second Y-axis: Conversion Rate
ax2 = ax1.twinx()

ax2.plot(
    months,
    conversion_rate,
    marker="s",
    linestyle="--",
    label="Conversion Rate"
)

ax2.set_ylabel("Conversion Rate (%)")

plt.title("Website Performance Analysis")

plt.show()
```
