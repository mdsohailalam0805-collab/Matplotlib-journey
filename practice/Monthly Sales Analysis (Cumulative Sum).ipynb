```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = np.array([10000, 15000, 12000, 18000, 20000, 25000])
```

```python
# Calculate cumulative sales
cumulative_sales = np.cumsum(sales)

plt.figure(figsize=(9, 5))

plt.plot(
    months,
    cumulative_sales,
    marker="o",
    linewidth=2,
    label="Cumulative Sales"
)

plt.title("Monthly Cumulative Sales")
plt.xlabel("Months")
plt.ylabel("Total Sales (₹)")
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

```python
print("Monthly Sales:", sales)
print("Cumulative Sales:", cumulative_sales)
print("Final Total Sales:", cumulative_sales[-1])
```
