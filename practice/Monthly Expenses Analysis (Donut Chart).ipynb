```python

import matplotlib.pyplot as plt

categories = ["Food", "Rent", "Travel", "Shopping", "Others"]
expenses = [5000, 10000, 3000, 4000, 2000]

plt.figure(figsize=(7, 7))

plt.pie(
    expenses,
    labels=categories,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.4}
)

plt.title("Monthly Expense Distribution")

plt.text(
    0, 0,
    f"Total\n₹{sum(expenses)}",
    ha="center",
    va="center",
    fontsize=14,
    fontweight="bold"
)

plt.show()

print("Total Expenses:", sum(expenses))
print("Highest Expense:", categories[expenses.index(max(expenses))])
```
