```python
import matplotlib.pyplot as plt 
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

revenue = [25000, 30000, 28000, 35000, 40000, 45000]
expenses = [18000, 20000, 22000, 25000, 28000, 30000]

profit = [revenue[i] - expenses[i] for i in range(len(months))]
```

```python
plt.figure(figsize=(9, 5))

plt.bar(months, revenue, label="Revenue")
plt.bar(months, expenses, label="Expenses")

plt.xlabel("Months")
plt.ylabel("Amount (₹)")
plt.title("Monthly Revenue vs Expenses")
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
print("Total Revenue:", sum(revenue))
print("Total Expenses:", sum(expenses))
print("Total Profit:", sum(profit))
print("Average Monthly Profit:", round(sum(profit) / len(profit), 2))
print("Highest Profit Month:", months[profit.index(max(profit))])
```
