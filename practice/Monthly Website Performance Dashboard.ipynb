```python
import matplotlib.pyplot as plt
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

visitors = [1200, 1500, 1300, 1800, 2200, 2500]
page_views = [3000, 3500, 3200, 4500, 5000, 6000]
bounce_rate = [45, 42, 48, 38, 35, 30]
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(months, visitors, marker="o", label="Visitors")
plt.plot(months, page_views, marker="s", label="Page Views")
plt.title("Website Traffic")
plt.xlabel("Months")
plt.ylabel("Count")
plt.legend()
plt.grid(True, alpha=0.3)
```
