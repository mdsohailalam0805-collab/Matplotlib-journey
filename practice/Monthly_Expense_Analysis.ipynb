```python
import matplotlib.pyplot as plt
```

```python
categories = ["Food", "Travel", "Shoping", "Bills", "Others"]
expenses = [700, 300, 1500, 3000, 5000]
```

```python
plt.pie(expenses , labels=categories, autopct= "%1.1f%%", startangle = 90)
plt.title("Monthly Expense Analysis")
plt.show()
```

```python
print("Total Expenses:", sum(expenses))
print("Highest Expense:", max(expenses))
print("Highest Spending Category:", categories[expenses.index(max(expenses))])
```
