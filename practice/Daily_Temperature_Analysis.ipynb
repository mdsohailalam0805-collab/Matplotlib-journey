```python

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

patna = [30, 32, 31, 34, 35, 33, 36]
delhi = [28, 31, 33, 35, 37, 36, 38]
mumbai = [29, 30, 31, 30, 32, 31, 33]

fig, axes = plt.subplots(
    1, 3,
    figsize=(13, 4),
    sharey=True
)

axes[0].plot(days, patna, marker="o")
axes[0].set_title("Patna")
axes[0].set_ylabel("Temperature (°C)")

axes[1].plot(days, delhi, marker="s", color="orange")
axes[1].set_title("Delhi")

axes[2].plot(days, mumbai, marker="^", color="green")
axes[2].set_title("Mumbai")

for ax in axes:
    ax.set_xlabel("Days")
    ax.grid(True, alpha=0.3)

fig.suptitle("Weekly Temperature Comparison")
plt.tight_layout()
plt.show()

print("Patna Average:", round(sum(patna) / len(patna), 2))
print("Delhi Average:", round(sum(delhi) / len(delhi), 2))
print("Mumbai Average:", round(sum(mumbai) / len(mumbai), 2))
```
