```python
import matplotlib.pyplot as plt

students = ["Aman", "Rahul", "Sohail", "Priya", "Neha", "Riya"]
marks = [78, 85, 92, 67, 88, 74]
```

```python
plt.bar(students, marks, color="orange")

plt.axhline(sum(marks) / len(marks), linestyle="--", label="Average Marks")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.title("Student Marks Analysis")
plt.legend()
plt.show()
```

```python
print("Average Marks:", round(sum(marks) / len(marks), 2))
print("Highest Marks:", max(marks))
print("Topper:", students[marks.index(max(marks))])
print("Lowest Marks:", min(marks))
```
