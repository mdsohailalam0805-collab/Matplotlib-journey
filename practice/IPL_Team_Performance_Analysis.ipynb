```python
import matplotlib.pyplot as plt
```

```python
teams = ["CSK", "MI", "RCB", "KKR", "RR"]
wins = [6, 8, 5, 7, 4]
```

```python
plt.figure(figsize=(8, 5))

plt.bar(teams, wins, color="orange", edgecolor="black")

plt.xlabel("Teams")
plt.ylabel("Matches Won")
plt.title("IPL Team Performance")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```

```python
print("Total Wins:", sum(wins))
print("Most Wins:", max(wins))
print("Top Team:", teams[wins.index(max(wins))])
print("Average Wins:", round(sum(wins) / len(wins), 2))
```
